**Ejercicio 2: La Calculadora de KPIs Necesitas calcular la ganancia neta de una venta restando los costos operativos.
Crea una función llamada calcular_ganancia que reciba dos parámetros: ingreso_total y costos. La función debe calcular la resta de estos dos valores y retornar (return) el resultado. Finalmente, prueba tu función con ingresos y costos insertados por consola, guardando el resultado en una variable e imprimiéndolo.**


```python
def calcular_ganancia(ingreso_total, costos):
    resta = ingreso_total - costos
    return resta

solicitud = int(input('dame el ingreso total: '))
costos = int(input('dame los costos: '))
resta = calcular_ganancia(solicitud, costos)
print(resta)
```

**Ejercicio 3: El Estandarizador de Columnas (Limpieza de texto) Los nombres de las columnas que vienen de Excel suelen venir en mayúsculas y con espacios, lo que dificulta el análisis. 
Crea una función llamada limpiar_columna que reciba una cadena de texto (texto_sucio). La función debe convertir el texto a minúsculas y reemplazar los espacios " " por guiones bajos "_" usando el método .replace(). Retorna el texto limpio. Pruébala con el texto que quieras tomándolo con método input**


```python
def limpiar_columna (texto_sucio):
    # convierte a minusculas y elimina espacios al inicio y al final:
    texto_limpio = texto_sucio.lower().strip().replace( "   ", "_") # siempre se aplican en la variable
    return texto_limpio

# uso 
columna_sucia = input('escribe el texto que quieras: ')
columna_estandar = limpiar_columna (columna_sucia)

print(f'Columna lista para la base de datos: {columna_estandar}')
```

**Ejercicio 5: El Procesador de Lotes (Integrando bucles)
Tienes una lista de temperaturas extraídas de un sensor y necesitas saber cuántas de ellas superan los 30 grados para lanzar una alerta. Instrucción: Crea una función llamada contar_alertas que reciba una lista_temperaturas. La función debe inicializar un contador en 0, usar un ciclo for para recorrer la lista, y si la temperatura es mayor a 30, sumar 1 al contador. Al final, debe retornar el total de alertas. Pruébala con la lista [25, 32, 28, 35, 31, 22].**

```python
def contar_alertas (lista_temperaturas):
    contar = 0
    for temperatura in lista_temperaturas:
        if temperatura > 30:
            contar += 1
    return contar  
print(contar_alertas([25, 32, 28, 35, 31, 22]))     
```

**Ejercicio 6: El Filtro de Calidad
Al extraer datos de un sistema antiguo, algunas edades de clientes se registraron con números negativos o ceros por error. Necesitamos auditar eso.
Crea una función llamada auditar_edades que reciba una lista de números (lista_edades). La función debe iniciar un contador de errores en 0, usar un ciclo for para recorrer cada edad, y un condicional if para evaluar si la edad es menor o igual a 0. Si lo es, suma 1 al contador de errores. Al final, la función debe retornar el total de errores encontrados. Pruébala con la lista [25, -2, 30, 0, 45, 18,-5 , 0 , 52].**

```python
def auditar_edades (lista_edades):
    contar = 0
    for edad in lista_edades:
        if edad <= 0:
            contar += 1
    return contar

print(auditar_edades([25, -2, 30, 0, 45, 18,-5 , 0 , 52])) 
```

**Reto Extra 2: El Generador de Correos (Función que inserta en listas)
El equipo de marketing te pasó una lista con nombres de empleados nuevos y necesitas crearles sus correos corporativos automáticamente y guardarlos en un nuevo registro. Crea una función llamada generar_correos que reciba una lista de nombres (lista_nombres). Dentro de la función, crea una lista vacía llamada lista_correos. Usa un ciclo for para recorrer los nombres; a cada nombre agrégale la terminación "@empresa.com" y conviértelo a minúsculas. Usa el método .append() para insertar este nuevo correo en tu lista vacía. Al final, retorna la lista completa de correos. Pruébala con valores dados con input.**


```python

def generar_correos (lista_nombres):
    lista_correos = []
    
    for nombre in lista_nombres:
        correo_empresa = nombre.lower() + '@empresa.com'
        lista_correos.append(correo_empresa)

    return lista_correos 
lista_nombres = input('Ingrese el nombre de los empleados separados por comas: ').split(',')
print(generar_correos(lista_nombres))


   
```
