**Ejercicio 1: Evaluación de Promoción Comercial (Comercio / Marketing)** 

Contexto: Una tienda en línea desea evaluar si aplica un descuento especial a sus clientes según el monto total de su compra y su categoría (Bronce, Plata, Oro).  

• Solicitar la cantidad de clientes a procesar.  
•Procesar a cada cliente solicitando: Nombre, Categoría (1: Bronce, 2: Plata, 3: Oro) y Monto total de la compra.  
• Determinar el porcentaje de descuento: • Categoría Oro (3): 20% de descuento si la compra es mayor a $100; de lo contrario, 10%.  
• Categoría Plata (2): 10% de descuento si la compra es mayor a $100; de lo contrario, 5%.   
• Categoría Bronce (1): No recibe descuento (0%).  
• Calcular el monto final a pagar (Monto - Descuento) y mostrar el resumen detallado para cada cliente.  
• Mostrar el promedio de compras por categoría (Oro, Plata, Bronce).

**Ejercicio 2: Clasificación de Riesgo en Salud (Salud)**

Contexto: Un centro médico requiere un sistema para evaluar a un conjunto de pacientes, calcular su Índice de Masa Corporal (IMC) y generar una alerta preventiva para aquellos que presenten una combinación de alto riesgo.  
• Solicitar el número de pacientes a evaluar.  
• Solicitar por cada paciente: Nombre, Peso (kg), Estatura (m) y Edad.  
• Calcular el IMC con la fórmula: IMC = peso / (estatura ^ 2).  
• Clasificar la categoría del paciente usando condicionales anidados: • IMC < 18.5: Bajo peso • 18.5 <= IMC < 25.0: Peso normal • 25.0 <= IMC < 30.0: Sobrepeso • IMC >= 30.0: Obesidad  
• Si el paciente tiene Obesidad y es mayor de 60 años, emitir el mensaje de alerta: 'Requiere atención prioritaria por geriatría'.  
• Imprimir el diagnóstico y alertas correspondientes por cada paciente.  
• Mostrar el total de pacientes en cada categoría

**Ejercicio 3: Sistema de Gestión Académica y Becas (Educación)**

Contexto: Una institución educativa desea evaluar el rendimiento de un grupo de estudiantes durante un periodo lectivo para calcular sus promedios e identificar a los candidatos a una Beca de Honor.  
• Solicitar la cantidad total de estudiantes a evaluar.  
• Usar un ciclo principal para iterar sobre cada estudiante (solicitando Nombre y Cantidad de materias cursadas).  
• Utilizar un ciclo anidado para solicitar la nota final de cada materia (notas entre 0.0 y 5.0) y acumular el total para calcular el promedio final.  
• Evaluar la condición de aprobación: Promedio >= 3.0 es 'Aprobado', de lo contrario 'Reprobado'.  
• Evaluar la asignación de beca: Un estudiante obtiene la 'Beca de Honor' si su Promedio >= 4.5 Y no reprobó ninguna materia individualmente (todas sus notas fueron >= 3.0).  
• Imprimir el boletín detallado con promedio, estado final y condición de beca. 
• Mostrar al final cuántos estudiantes ganaron y cuantos perdieron.
