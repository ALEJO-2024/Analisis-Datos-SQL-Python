6.	El Liquidador de Comisiones (Lógica Combinada)
Contexto: Eres el analista encargado de calcular si un vendedor recibe comisión este mes, combinando varias reglas de negocio. Instrucción: Solicita las ventas totales del mes (número) y la antigüedad del vendedor en años (número).
•	Si las ventas superan los 10000 y la antigüedad es mayor a 2 años, gana comisión del 10%.
•	Si las ventas superan los 10000 pero tiene 2 años o menos, o si las ventas están entre 5000 y 10000 sin importar la antigüedad, gana comisión del 5% (usa elif).
•	Si no cumple nada de lo anterior, no hay comisión. Muestra el mensaje correspondiente.


```python
comision = float(0)

ventas_totales = int(input("Ingrese el total de ventas: "))
antiguedad=int(input("Ingrese la antigüedad del vendedor en años: "))

if ventas_totales > 10000 and antiguedad >= 2:
    comision = ventas_totales * 0.10
    print(f"Comisión calculada: {comision}")
elif (ventas_totales > 10000 and antiguedad <= 2) or (ventas_totales > 5000 and ventas_totales <= 10000):
    comision = ventas_totales * 0.05
    print(f"Comisión calculada: {comision}")
else:
    print("No se aplica comisión")
```
