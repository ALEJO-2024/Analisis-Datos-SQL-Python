Crea un programa en Python que tenga una variable llamada **temperatura**.

El programa debe evaluar:

* Si la temperatura es mayor o igual a 30, mostrar: **"Hace mucho calor".**
* Si la temperatura es mayor o igual a 20 pero menor que 30, mostrar: **"El clima es agradable".**
* En cualquier otro caso, mostrar: **"Hace frío".**


```python
temperatura = int(input('Ingresse la temperatura actual: '))
llevar_paraguas = False

if temperatura >= 30:
    print('Hace mucho calor')
    llevar_paraguas = False
elif temperatura > 20 and temperatura < 30:
    print('El clima es agradable')
    llevar_paraguas = True
else: 
    print('Hace frio')
    llevar_paraguas = True
    print('Debo ir con paraguas? ', llevar_paraguas)


```

Crea un programa en Python que evalúe la edad y la ciudad de residencia de una persona utilizando operadores de comparación y operadores lógicos.
 
El programa debe:
 
Definir dos variables:
edad: un número entero que representa la edad de la persona.
ciudad: una cadena de texto que representa la ciudad donde vive.
 
Evaluar las siguientes condiciones utilizando if, elif y else:
 
Si la persona tiene 18 años o más y vive en Medellín, mostrar el mensaje:
"Mayor de edad y vive en Medellín".
 
Si la persona tiene 18 años o más pero vive en otra ciudad distinta de Medellín, mostrar el mensaje:
"Mayor de edad pero vive en otra ciudad".
 
Si la persona es menor de edad o vive en Medellín, mostrar el mensaje:
"Menor de edad o vive en Medellín".
 
En cualquier otro caso, mostrar el mensaje:
"No cumple ninguna condición".
 
Objetivo:
Practicar el uso de operadores lógicos (and, or), operadores de comparación (>=, <, ==, !=) y estructuras de control if, elif, else en Python.

```python
edad = 18
ciudad = 'medellin'

solicitud_edad = int(input('Ingrese su edad: '))
solicitud_ciudad = input("ingrese la ciudad de residencia: ").lower().strip().replace("í", "i")


if solicitud_edad >= 18 and solicitud_ciudad == 'medellin':
    print('Mayor de edad y vive en Medellín')

elif solicitud_edad >= 18 and solicitud_ciudad != 'medellin':
    print('Mayor de edad pero vive en otra ciudad')

elif solicitud_edad < 18 or solicitud_ciudad == 'medellin':
    print('Menor de edad o vive en Medellín')

else: print('No cumple ninguna condición')
```
