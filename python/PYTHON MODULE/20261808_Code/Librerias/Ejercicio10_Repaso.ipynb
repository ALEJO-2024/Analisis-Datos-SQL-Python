**Ejercicio 24: Agregaciones Múltiples con .agg()**  

•	Contexto: El departamento financiero requiere un reporte avanzado que detalle simultáneamente el ingreso total y el precio promedio por categoría.  

•	Instrucción: Agrupa el DataFrame por "categoria" y utiliza el método .agg() para calcular la suma de "venta_total" y el promedio de "precio". Restablece el índice con .reset_index() y muestra la tabla resultante.


```python
departamento_financiero = df.groupby('categoria').agg(
    venta_total = ('venta_total', 'sum'), 
    valor = ('precio', 'mean'))

departamento_financiero = departamento_financiero.reset_index()
departamento_financiero
```

**Ejercicio 30: Consolidación y Resumen Final con Funciones y Bucles**  

•	Contexto: Crearemos un bloque de script que combine lógica de funciones personalizadas y bucles para resumir estadísticas clave del dataset limpio.  

•	Instrucción: Define una función que reciba un DataFrame y devuelva el ingreso total acumulado ("venta_total"). Utiliza un bucle for para iterar sobre las ciudades únicas del DataFrame, imprimiendo el ingreso total generado en cada ciudad específica calculado mediante un filtro dinámico.



```python
def calcular_ingreso_total(df_filtrado):
    # Sumamos los valores de la columna "venta_total"
    ingreso_total = df_filtrado['venta_total'].sum()
    return ingreso_total

# Esto nos da una lista con los nombres de cada ciudad única
ciudades_unicas = df['ciudad'].unique()

# Iteramos sobre cada ciudad de la lista
for ciudad in ciudades_unicas:
    # FILTRO DINÁMICO: Seleccionamos solo las filas de la ciudad actual
    df_ciudad = df[df['ciudad'] == ciudad]
    
    # LLAMADA A LA FUNCIÓN: Calculamos el ingreso total de esa ciudad
    total_ciudad = calcular_ingreso_total(df_ciudad)
    
    # IMPRESIÓN: Mostramos el resultado en pantalla
    print(f"El ingreso total generado en la ciudad de {ciudad} es: ${total_ciudad:.2f}")
```
