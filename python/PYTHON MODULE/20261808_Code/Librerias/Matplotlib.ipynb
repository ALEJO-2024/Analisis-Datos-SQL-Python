```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
meses = ['Enero', 'Febrero', 'Marzo', 'Abril', 'Mayo', 'Junio'] # lista o set de datos
ventas = [1500, 2000, 1800, 2200, 2500, 3000] # lista o set de datos
plt.figure(figsize=(8, 6)) # tamaño de la figura
plt.title('Ventas Mensuales', fontsize=16) # título del gráfico
plt.xlabel('Meses', fontsize=14) # etiqueta del eje x
plt.ylabel('Ventas ($)', fontsize=14) # etiqueta del eje y
plt.grid(True, linestyle=':', alpha=0.9, linewidth=0.5) # agregar cuadrícula
plt.plot(meses, ventas, marker='o', linestyle=':', color='b', linewidth=1)

plt.show()
```

```python
# GRAFICO DE BARRAS

categorias = ['Accesorios', 'Ropa', 'Calzado', 'Electrónica', 'Hogar'] # lista o set de datos
unidades = [120, 250, 180, 300, 150] # lista o set de datos
plt.figure(figsize=(8, 6)) # tamaño de la figura
plt.title('Unidades Vendidas por Categoría', fontsize=16) # título del gráfico
plt.bar(categorias, unidades, color='aqua', alpha=0.7) # gráfico de barras
plt.grid(True, linestyle=':', alpha=0.9, linewidth=0.5) # agregar cuadrícula
```

```python
# DIAGRAMA DE DISPERSION

precios = [50000, 60000, 55000, 70000, 65000] # lista o set de datos
cantidades = [10, 15, 12, 20, 18] # lista o set de datos

plt.style.use('dark_background') # estilo del gráfico
plt.figure(figsize=(8, 6)) # tamaño de la figura
plt.title('Relación entre Precios y Cantidades', fontsize=16) # título del gráfico
plt.xlabel('Precios ($)', fontsize=14) # etiqueta del eje x
plt.ylabel('Cantidades', fontsize=14) # etiqueta del eje y
plt.scatter(precios, cantidades, color='lightseagreen', alpha=0.7) # diagrama de dispersión
plt.grid(True, linestyle=':', alpha=0.9, linewidth=0.5) # agregar cuadrícula
plt.show()
```

```python

# Generacion de datos aleatorios para el histograma
precios_dataset = np.random.normal(50000, 20000, 1000) # datos aleatorios con distribución normal

plt.hist(precios_dataset, rwidth=0.8, bins=30, color='lightseagreen', alpha=0.7) # histograma
plt.xlabel('Precios ($)', fontsize=14) # etiqueta del eje x
plt.ylabel('Frecuencia', fontsize=14) # etiqueta del eje y
plt.title('Distribución de Precios', fontsize=16) # título del gráfico
plt.grid(True, linestyle=':', alpha=0.9, linewidth=0.5) # agregar cuadrícula
plt.show()
```

```python
#SUBPLOTS (MULTIPLES GRAFICOS EN UNA SOLA FIGURA)

import matplotlib.pyplot as plt
plt.style.use('ggplot')
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(8, 4))
 
# Gráfico 1 (Izquierda)
ax1.plot([1, 2, 3], [10, 20, 25], color='blue')
ax1.set_title('Tendencia A')
 
# Gráfico 2 (Derecha)
ax2.bar(['A', 'B', 'C'], [5, 12, 8], color='orange')
ax2.set_title('Comparativa B')
 
plt.tight_layout()
plt.show()
```
