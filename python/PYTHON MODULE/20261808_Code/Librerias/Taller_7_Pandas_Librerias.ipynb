Ejercicio 2: Creando una Serie en Pandas  

•	Contexto: Tienes un listado de precios de productos y quieres organizarlos con sus respectivas etiquetas de categoría.  

•	Instrucción: Importa pandas as pd. Crea una lista con los precios [1200, 450, 89] y una lista de etiquetas ["Laptop", "Mouse", "Teclado"] para usarlas como índice. Convierte esto en una pd.Series e imprímela.


```python
import pandas as pd

lista_precios = [1200, 450, 89] 
lista_etiquetas = ["Laptop", "Mouse", "Teclado"] 

# convierto estas listas a DataFrame:
df_lista_precios = pd.DataFrame(lista_precios, lista_etiquetas)


print('--- DataFrame desde listas ---')

print(df_lista_precios)
```

Ejercicio 3: Creando un DataFrame desde un Diccionario  

•	Contexto: Necesitas armar una tabla rápida con información de sucursales para una presentación ejecutiva.  

•	Instrucción: Importa pandas as pd. Crea un diccionario con dos columnas: "Ciudad" (["Bogotá", "Medellín", "Cali"]) y "Ventas" ([5000, 4200, 3100]).  
Convierte el diccionario en un DataFrame y muéstralo en consola.

```python
import pandas as pd

ciudad = (["Bogotá", "Medellín", "Cali"])
ventas = ([5000, 4200, 3100]) 

# convierto estos diccionarios a DataFrame:
df_ciudad = pd.DataFrame({'Ciudad': ciudad, 'Ventas': ventas})


print('--- DataFrame desde diccionario ---')

print(df_ciudad)
```

Ejercicio 4: DataFrame desde una Lista de Diccionarios (Formato API)  

•	Contexto: Recibiste datos de transacciones en formato de diccionarios independientes y debes estructurarlos en una tabla limpia.  

•	Instrucción: Importa pandas as pd. Crea una lista con tres diccionarios, cada uno con las claves "ID" y "Estado". Conviértela en un DataFrame e imprímela.


```python
import pandas as pd

# lista con 3 diccionarios con las claves "ID" y "Estado":

datos = [
    {"ID": 1, "Estado": "Completada"},
    {"ID": 2, "Estado": "Pendiente"},
    {"ID": 3, "Estado": "Cancelada"}
]


# convierto estos diccionarios a DataFrame:
df_datos = pd.DataFrame(datos)


print('--- DataFrame desde diccionario Formato API---')

print(df_datos)
```
