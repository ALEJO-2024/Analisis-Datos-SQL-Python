1.	Validador de Asistencia: Solicita al usuario ingresar el estado de registro de un asistente (debe ser estrictamente "presente" o "ausente"). El ciclo debe repetirse mientras el texto ingresado no sea válido.

```python
asistencia = input('Ingresa el estado el usuario "presente / ausente": ')
 
while asistencia != 'presente' and asistencia != 'ausente':
    print('Texto ingresado no valido')
    asistencia = input('Ingresa el estado el usuario "presente / ausente": ')
 
print(f'el usuario tiene una asistencia {asistencia}')
```

2.	Límite de Reintentos de Conexión: Simula un inicio de sesión a una base de datos. Pide una contraseña y otorga un máximo de 3 intentos. El ciclo continúa mientras la clave sea incorrecta y la cantidad de intentos sea menor a 3.

```python

password_real = 12345

i = 1 # Inicio o Start

while i <= 3: # Final o Stop
    
    password = int(input('Ingrese su contraseña: '))

    if password == password_real:
            print('Contraseña correcta')
            break
    
    elif password != password_real:
        print('Contraseña incorrecta, ingrese su contraseña de nuevo: ')
        i = i + 1 # Paso o Step

    if i > 3:        

        print('supero el limite de intentos')

   
```

3.	Capacidad del Evento Un salón tiene una capacidad estricta de 50 personas. Pide la cantidad de personas por grupo que van ingresando. El ciclo suma los asistentes mientras el total sea menor a 50 o el grupo ingresado sea mayor o igual a 1. Si supera el aforo, muestra el total.

```python
cantidad_ingresos = int(input('Ingresa la cantidad de personas por grupo: '))
capacidad = 50

for i in range(1, 51):
    
    if cantidad_ingresos < capacidad:
        print('aún hay cupo disponible')

        capacidad = capacidad - cantidad_ingresos
        print(f'Capacidad restante: {capacidad}')

        cantidad_ingresos = int(input('Ingresa la cantidad de personas por grupo: '))

    elif cantidad_ingresos == capacidad:
        print('aforo completo: ', 50)
        break
           
```

5.	Limpieza de Errores con continue Tienes una lista de transacciones donde algunos valores fallaron y se guardaron como 0 o -1. Recorre la lista; si el valor es 0 o -1, sáltalo. Si es correcto, imprímelo.
transacciones = [150, 0, 320, -1, 400,250,482]


```python
transacciones = [150, 0, 320, -1, 400,250,482]

for valor in transacciones:
    if valor == 0 or valor == -1:
        continue 
    print(valor) # Salta el resto del código y va a la siguiente iteración
   
```

tiempos_ms = [15, 20, 105, 30,22]
