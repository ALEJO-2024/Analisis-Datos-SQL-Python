1.	Calculadora de Edad (Calentamiento) Escribe un programa que pregunte al usuario su año de nacimiento y el año actual. El programa debe calcular su edad y mostrar el siguiente mensaje en pantalla: "Tienes [edad] años o los cumplirás este año".

```python
año_nacimiento = int(input('ingresa tu año de nacimiento'))
año_actual = int(input('ingresa el año actual'))
edad = (año_actual - año_nacimiento)
print(f"Tienes {edad} años o los cumplirás este año")
```

2.	Promedio de Satisfacción al Cliente (CX) Imagina que estás procesando los resultados de un formulario de encuestas de servicio al cliente. El programa debe pedir al usuario que ingrese las calificaciones de 3 encuestas diferentes (números enteros del 1 al 5). Calcula el promedio exacto de estas tres calificaciones (con decimales) y muéstralo en pantalla

```python
calificacion1 = float(input('ingrese la calificacion numero 1 del 1 al 5'))
calificacion2 = float(input('ingrese la calificacion numero 2 del 1 al 5'))
calificacion3 = float(input('ingrese la calificacion numero 3 del 1 al 5'))
promedio_satisfaccion = (calificacion1 + calificacion2 + calificacion3)/3
print ('el promedio de satisfaccion es: ' ,promedio_satisfaccion)
```

3.	Auditoría de Tiempos de Ejecución Tienes un script automatizado que se ejecuta en el servidor y tarda cierta cantidad de minutos en procesar la data. Construye un programa que pida al usuario ingresar el total de minutos que tardó el proceso. Utilizando los operadores de división entera (//) y módulo (%), convierte ese tiempo a horas y minutos. Ejemplo: Si el usuario ingresa 130, el programa debe imprimir: "El proceso tardó 2 horas y 10 minutos".



```python
tarda = int(input('ingresa la cantidad de minutos que tardo en procesar la data'))
calculo = tarda//60
minutos_restantes = tarda % 60
print (f"El proceso tardo : {calculo} horas y {minutos_restantes} minutos")
```
