Fundamentos y Dominio de Seaborn para Visualización Estadística

¿Qué es Seaborn? Es una librería de visualización de datos en Python construida por encima de Matplotlib. Está específicamente diseñada para facilitar la creación de gráficos estadísticos complejos, atractivos y de aspecto profesional con muy pocas líneas de código.

Integración con Pandas: A diferencia de Matplotlib, que trabaja directamente con arreglos o coordenadas sueltas, Seaborn entiende nativamente la estructura de los DataFrames de Pandas, permitiendo pasar nombres de columnas directamente a los ejes (x, y, hue).

Aplicaciones y Usos en el Mercado:

Análisis Exploratorio de Datos (EDA) Avanzado: Detección rápida de distribuciones, valores atípicos y relaciones multivariadas entre variables categóricas y numéricas.

Presentaciones Ejecutivas y Publicaciones Científicas: Ofrece paletas de colores estéticamente agradables y temas visuales listos para reportes de alta calidad sin necesidad de configuración manual extensa.

Ciencia de Datos y Machine Learning: Ideal para graficar matrices de correlación, curvas de densidad y distribuciones antes y después del entrenamiento de modelos.

Temas Principales para Dominar la Librería

Temas y Paletas de Colores: Configuración visual global con sns.set_theme() y uso de paletas (palette).

Gráficos Relacionales (scatterplot, lineplot): Visualización de relaciones continuas con la capacidad de agrupar por color (hue) o tamaño (size).

Gráficos Categóricos (boxplot, violinplot, countplot): Análisis de distribuciones y conteos en variables discretas.

Gráficos de Distribución (histplot, kdeplot): Representación de histogramas suavizados y densidades de probabilidad.

Gráficos Matriciales y Multivariados (heatmap, pairplot): Mapas de calor para matrices de correlación y visualización cruzada de múltiples variables en un solo vistazo.

Explicaciones Breves y Ejemplos Prácticos
1. Gráficos Relacionales de Dispersión (sns.scatterplot)

Explicación: Permite graficar la relación entre dos variables numéricas, añadiendo una tercera dimensión mediante color (hue) o tamaño para diferenciar categorías.

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Cargar dataset de ejemplo integrado en Seaborn
df = sns.load_dataset("tips")

sns.scatterplot(data=df, x="total_bill", y="tip", hue="time", style="smoker")
plt.title("Relación entre Total de Cuenta y Propina por Horario")
plt.show()
```

2. Diagramas de Caja (sns.boxplot)

Explicación: Resumen visualmente la distribución de una variable numérica a través de sus cuartiles, mediana y la identificación de valores atípicos (outliers) segmentados por categorías.

```python
import matplotlib.pyplot as plt
import seaborn as sns

# df = sns.load_dataset("tips")

sns.boxplot(data=df, x="day", y="total_bill", palette="Set2")
plt.title("Distribución de Cuentas por Día de la Semana")
plt.show()
```

3. Conteo de Categorías (sns.countplot)

Explicación: Cuenta automáticamente la frecuencia de observaciones en cada categoría de una variable discreta, funcionando como un histograma para datos categóricos.

```python
import matplotlib.pyplot as plt
import seaborn as sns

df = sns.load_dataset("titanic")

sns.countplot(data=df, x="class", hue="survived")
plt.title("Supervivientes por Clase en el Titanic")
plt.show()
```

4. Mapas de Calor (sns.heatmap)

Explicación: Representan datos matriciales (como una tabla de correlaciones) mediante una gradación de colores, facilitando la identificación inmediata de valores altos o bajos.

```python
import matplotlib.pyplot as plt
import seaborn as sns

df = sns.load_dataset("titanic")
# Calcular matriz de correlación numérica
corr = df.select_dtypes(include="number").corr()

sns.heatmap(corr, annot=True, cmap="coolwarm", fmt=".2f")
plt.title("Matriz de Correlación - Titanic")
plt.show()
```

5. Cruce Multivariable (sns.pairplot)

Explicación: Genera una cuadrícula de gráficos de dispersión e histogramas para todas las combinaciones numéricas de un DataFrame, ideal para un análisis exploratorio integral.

```python
import matplotlib.pyplot as plt
import seaborn as sns

df = sns.load_dataset("iris")

sns.pairplot(df, hue="species")
plt.show()
```
